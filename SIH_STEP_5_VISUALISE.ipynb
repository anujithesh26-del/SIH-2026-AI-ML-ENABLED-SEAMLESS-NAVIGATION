```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
import io

# Prompt user to upload cleaned_Vw10.csv
print("Please upload 'cleaned_Vw10.csv'")
uploaded_cleaned = files.upload()
cleaned_filename = list(uploaded_cleaned.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded_cleaned[cleaned_filename]))

# Prompt user to upload merged_Vw10.csv
print("Please upload 'merged_Vw10.csv'")
uploaded_merged = files.upload()
merged_filename = list(uploaded_merged.keys())[0]
raw = pd.read_csv(io.BytesIO(uploaded_merged[merged_filename]))

# Anomaly window identified in step 4 (Clean): rows where the V-file's
# satellite count was set to NaN for being out of range.
anomaly_mask = df["No of GPS Satellites Available"].isna()
t0 = df.loc[anomaly_mask, "time_s"].min()
t1 = df.loc[anomaly_mask, "time_s"].max()

fig, axes = plt.subplots(4, 1, figsize=(11, 13), sharex=True)

# --- Accelerometer XYZ (phone, S-file) ---
ax = axes[0]
for col, label in [
    ("S_ACCELEROMETER X (m/s²)", "X"),
    ("S_ACCELEROMETER Y (m/s²)", "Y"),
    ("S_ACCELEROMETER Z (m/s²)", "Z"),
]:
    ax.plot(df["time_s"], df[col], label=label, linewidth=0.8)
ax.axvspan(t0, t1, color="red", alpha=0.12, label="flagged anomaly window")
ax.set_ylabel("Accel (m/s²)")
ax.set_title("Phone accelerometer - raw")
ax.legend(loc="upper right", fontsize=8)

# --- Gyroscope Yaw/Pitch/Roll (phone, S-file) ---
ax = axes[1]
for col, label in [
    ("S_GYROSCOPE Yaw (rad/s)", "Yaw"),
    ("S_GYROSCOPE Pitch (rad/s)", "Pitch"),
    ("S_GYROSCOPE Roll (rad/s)", "Roll"),
]:
    ax.plot(df["time_s"], df[col], label=label, linewidth=0.8)
ax.axvspan(t0, t1, color="red", alpha=0.12)
ax.set_ylabel("Angular rate (rad/s)")
ax.set_title("Phone gyroscope - raw")
ax.legend(loc="upper right", fontsize=8)

# --- Vehicle velocity + wheel speeds (V-file) ---
ax = axes[2]
ax.plot(df["time_s"], df["Velocity (km/hr)"], label="GPS velocity", color="black", linewidth=1.2)
for col, label in [
    ("Wheel Speed Front Left (rad/sec)", "FL wheel"),
    ("Wheel Speed Front Right (rad/sec)", "FR wheel"),
]:
    ax.plot(df["time_s"], df[col], label=label, linewidth=0.6, alpha=0.6)
ax.axvspan(t0, t1, color="red", alpha=0.12)
ax.set_ylabel("Speed")
ax.set_title("Vehicle velocity vs. wheel speed - stays normal through the flagged window")
ax.legend(loc="upper right", fontsize=8)

# --- Satellite count anomaly itself, before it was set to NaN ---
ax = axes[3]
ax.plot(raw["time_s"], raw["No of GPS Satellites Available"], color="darkred", linewidth=0.9)
ax.axvspan(t0, t1, color="red", alpha=0.12)
ax.set_ylabel("Sat count")
ax.set_xlabel("time_s (seconds since start of day)")
ax.set_title("V-file 'No of GPS Satellites Available' - raw (pre-clean) - the anomaly this step is about")

plt.tight_layout()

output_imu_plot_name = "vw10_imu_traces.png"
plt.savefig(output_imu_plot_name, dpi=130)
files.download(output_imu_plot_name)
plt.close()

# --- GPS track (lat/lon) with anomaly window highlighted ---
fig, ax = plt.subplots(figsize=(7, 7))
normal = df.loc[~anomaly_mask]
anomaly = df.loc[anomaly_mask]
ax.plot(normal["Longitude (degrees)"], normal["Latitude (degrees)"],
        color="steelblue", linewidth=1.5, label="normal window")
ax.plot(anomaly["Longitude (degrees)"], anomaly["Latitude (degrees)"],
        color="red", linewidth=2.0, label="flagged anomaly window")
ax.scatter(df["Longitude (degrees)"].iloc[0], df["Latitude (degrees)"].iloc[0],
           color="green", zorder=5, label="start")
ax.scatter(df["Longitude (degrees)"].iloc[-1], df["Latitude (degrees)"].iloc[-1],
           color="black", zorder=5, label="end")
ax.set_xlabel("Longitude (degrees)")
ax.set_ylabel("Latitude (degrees)")
ax.set_title("GPS track - Vw10 session (V-file lat/lon)")
ax.legend()
ax.set_aspect("equal", adjustable="datalim")
plt.tight_layout()

output_gps_plot_name = "vw10_gps_track.png"
plt.savefig(output_gps_plot_name, dpi=130)
files.download(output_gps_plot_name)
plt.close()

print(f"Anomaly window highlighted: time_s {t0} to {t1} ({anomaly_mask.sum()} rows)")
```
